# NileRed BZ Virtual Twin v0.2

This notebook regenerates the interactive browser demo from the normalized Oregonator reaction–diffusion model. It shows a virtual vessel cross-section and a model signal. The colors are an abstract `x`-state proxy, not measured concentrations or calibrated video pixels. The external source video is included only as context; synchronization remains a future calibration gate.

```python
from pathlib import Path
from IPython.display import IFrame, display
from nilered_twin import OregonatorParameters, SpatialOregonatorParameters, simulate_oregonator, simulate_oregonator_1d
from nilered_twin.virtual_twin import write_virtual_twin_bundle

OUTPUTS = Path("../outputs")
kinetics = OregonatorParameters()
well_mixed = simulate_oregonator(kinetics)
spatial = simulate_oregonator_1d(SpatialOregonatorParameters(kinetics=kinetics))
bundle = write_virtual_twin_bundle(OUTPUTS, well_mixed=well_mixed, spatial=spatial)
print("VIRTUAL TWIN BUNDLE")
print("html:", OUTPUTS / bundle["html"])
print("data:", OUTPUTS / bundle["data"])
print("frames:", bundle["frames"])
print("calibration status:", bundle["calibration_status"])
print("well-mixed peaks:", well_mixed["peak_count"])
print("mean period:", f'{well_mixed["mean_period"]:.6f}', "dimensionless model units")
print("spatial solver success:", spatial["solver_success"])
display(IFrame(src="../outputs/virtual_twin.html", width="100%", height=920))
```
